Variable scope = where a variable is visible and accessible
 *Scope resolution = (LEGB): LOCAL -> ENCLOSED -> GLOBAL -> BUILT-IN    

```python
# LOCAL

def func1():
    x = 9
    print(x)

func1()
```

```python
# ENCLOSED

def func1():
    x = 7
    
    def func2():
        print(x)

    func2()

func1()
```

```python
# GLOBAL
x = 4
def func1():
    print(x)

func1()
```

```python
# BUIL-IN
from math import e
#But what if we already have a variable named e?
# e = 2     -> now 'e' has changed to 2
def func1():
    print(e)

func1()
```

```python
#Guessing test 1
x = "global"

def outer():
    x = "enclosing"

    def inner():
        x = "local"
        print(x)

    inner()
    print(x)

outer() # local, enclosing
print(x) # global
```

```python
def outer():
    x = "enclosing"
    def inner():
        nonlocal x
        x = "changed"
    inner()
    print(x)   # changed
```

```python
    #Guessing test 2
    count = 0

    def outer():
        count = 10

        def inner():
            nonlocal count      # This mean the enclosed variable(count = 10)
            count += 1          

        inner()                 # a single call -> 11 
        print(count)            # 11

    def reset():
        global count            
        count = 100

    outer()                     # 11     
    print(count)                # 0 
    reset()                     # 'Count' changed to 100
    print(count)                # 100
```
