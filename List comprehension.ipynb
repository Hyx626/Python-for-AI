# List conprehension = A concise way to create list in Python
#                    compact and easier to read than tranditional loops
#                    [exprssion for value in interable if condition]

```python
#[exprssion for value in interable if condition]
doubles = [x * 2 for x in range(1, 11)]
triples = [e * 3 for e in range(1, 11)]
squares = [y ** y for y in range(1, 11)]
```

```python
fruits = ["apple", "banana", "orange", "coconut"]
fruits = [fruit.upper() for fruit in fruits]
print(fruits)
```

```python
numbers = [1,3,-5,7,4,-6]

positive_nums = [num for num in numbers if num > 0]
negative_nums = [num for num in numbers if num < 0]
even_nums = [num for num in numbers if num % 2 == 0]

print(positive_nums)
print(negative_nums)
print(f'this is even numbers {even_nums}')
```

```python
numbers = [1, 3, -5, 7, 4, -6]

# Viết list comprehension → [1, 9, 49, 16]
square_positive_nums = [num * num for num in numbers if num > 0]

# biến số âm thành 0 → [1, 3, 0, 7, 4, 0]
replace_negative_nums = [max(num, 0) for num in numbers]

print(square_positive_nums)
print(replace_negative_nums)
```

```python
words = ["blender", "ai", "python", "film"]

# ['BLENDER', 'PYTHON', 'FILM']
upper_word = [word.upper() for word in words if len(word) > 3]

print(upper_word)
```
