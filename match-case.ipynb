# Math - case statement (switch): An altenative to ussing many 'elif' statements
 - Execute some code if a calue mathces a 'case'
 - Benefits: clearer and syntax is more readable
 

```python
# Return day of week
def day_of_week(day):
    match day:
        case 1:
            return "It's Sunday"
        case 2:
            return "It's Monday"
        case 3:
            return "It's Tuesday"
        case 4:
            return "It's Wednesday"
        case 5:
            return "It's Thursday"
        case 6: 
            return "It's Friday"
        case 7:
            return "It's Saturday"    
        case _:                     # "_" = wild card
            return "Not a valid day"

print(day_of_week(4))
```

```python
# It is weekend?
def is_weekend(day):
    match day:
        case "Sunday":
            return True
        case "Monday":
            return False
        case "Tuesday":
            return False 
        case "Wednesday":
            return False
        case "Thursday":
            return False
        case "Friday": 
            return False
        case "Saturday":
            return True
        case _:                     # "_" = wild card
            return False

print(is_weekend("Monday"))
```

```python
# It is weekend?
def is_weekend(day):
    match day:
        case "Sunday" | "Saturday":
            return True
        case _:                     # "_" = wild card
            return False

print(is_weekend("Saturday"))
```

```python
# Viết hàm grade(score) dùng match-case, nhận điểm chữ rồi trả về nhận xét:
# "A" → "Excellent"
# "B" hoặc "C" → "Good"
# "D" → "Pass"
# còn lại → "Fail"

def grade(score):
    match score.upper():
        case "A":
            return "Excellent"
        case "B" | "C":
            return "Good"
        case "D":
            return "Pass"
        case _:
            return "Fail"

print(grade("c"))
```
